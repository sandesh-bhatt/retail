```python
import pandas as pd
import requests
from io import BytesIO
from zipfile import ZipFile

url = "https://archive.ics.uci.edu/static/public/352/online+retail.zip"

z = ZipFile(BytesIO(requests.get(url).content))
df = pd.read_excel(z.open(z.namelist()[0]))

print(df.head())
```

StockCode	ID	:Categorical	a 5-digit integral number uniquely assigned to each distinct product

```python
df_clean = df.drop_duplicates().copy()

print("Original rows:", len(df))
print("Rows after removing duplicates:", len(df_clean))
```

```python
df_clean['Description'] = (
    df_clean['Description'].fillna('Unknown')
)
```

```python
df_clean.info()
```

```python
df_clean['InvoiceNo'] = df_clean['InvoiceNo'].astype(str)

cancelled = df_clean[
    df_clean['InvoiceNo'].str.startswith('C')
].copy()

negative_quantity = df_clean[
    df_clean['Quantity'] < 0
].copy()

print("Cancellation rows:", len(cancelled))
print("Negative quantity rows:", len(negative_quantity))
```

```python
sales_df = df_clean[
    (~df_clean['InvoiceNo'].str.startswith('C')) &
    (df_clean['Quantity'] > 0) &
    (df_clean['UnitPrice'] > 0)
].copy()
```

```python
sales_df['Revenue'] = (
    sales_df['Quantity'] * sales_df['UnitPrice']
)

sales_df.head()
```

```python
print("Cleaned dataset shape:", df_clean.shape)
print("Sales dataset shape:", sales_df.shape)

print("\nRemaining missing values:")
print(df_clean.isna().sum())

print("\nRemaining exact duplicates:")
print(df_clean.duplicated().sum())

print("\nSales revenue:")
print(sales_df['Revenue'].sum())
```

```python
df_clean.to_csv(
    'online_retail_cleaned.csv',
    index=False
)

sales_df.to_csv(
    'online_retail_sales.csv',
    index=False
)
```

```python
sales_df['InvoiceDate'] = pd.to_datetime(
    sales_df['InvoiceDate']
)

sales_df['Month'] = (
    sales_df['InvoiceDate'].dt.to_period('M')
)

monthly_revenue = (
    sales_df.groupby('Month')['Revenue']
    .sum()
    .sort_index()
)

monthly_revenue
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(14, 6))

monthly_revenue.plot(kind='line', marker='o')

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (£)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
print("Highest revenue month:", monthly_revenue.idxmax())
print("Highest monthly revenue:", monthly_revenue.max())

print("Lowest revenue month:", monthly_revenue.idxmin())
print("Lowest monthly revenue:", monthly_revenue.min())
```

```python
top_products = (
    sales_df.groupby('Description')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title('Top 10 Best-Selling Products by Quantity')
plt.xlabel('Total Quantity Sold')
plt.ylabel('Product')
plt.tight_layout()
plt.show()
```

```python
customer_frequency = (
    sales_df.dropna(subset=['CustomerID'])
    .groupby('CustomerID')['InvoiceNo']
    .nunique()
    .sort_values(ascending=False)
)

customer_frequency.head(10)
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    customer_frequency,
    bins=30
)

plt.title('Distribution of Customer Purchase Frequency')
plt.xlabel('Number of Unique Invoices')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()
```

```python
print("Most frequent customers:")
print(customer_frequency.head(10))

print("\nAverage invoices per customer:")
print(customer_frequency.mean())

print("\nCustomers with repeat purchases:")
print((customer_frequency > 1).sum())
```

```python
country_revenue = (
    sales_df.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

country_revenue
```

```python
top_countries = country_revenue.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title('Top 10 Countries by Revenue')
plt.xlabel('Revenue (£)')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
print("Highest revenue country:", country_revenue.idxmax())
print("Revenue:", country_revenue.max())
```

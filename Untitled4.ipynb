```python
customer_data = sales_df.dropna(subset=['CustomerID']).copy()

customer_data['TotalPrice'] = (
    customer_data['Quantity'] * customer_data['UnitPrice']
)
```

```python
customer_features = customer_data.groupby('CustomerID').agg(
    TotalSpent=('TotalPrice', 'sum'),
    TotalQuantity=('Quantity', 'sum'),
    TotalInvoices=('InvoiceNo', 'nunique'),
    AverageOrderValue=('TotalPrice', 'mean')
)

customer_features.head()
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

X = np.log1p(customer_features)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.xticks(range(2, 11))
plt.grid(True)
plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

customer_features['Cluster'] = kmeans.fit_predict(X_scaled)

customer_features.head()
```

```python
cluster_summary = customer_features.groupby('Cluster').agg(
    NumberOfCustomers=('TotalSpent', 'size'),
    AverageSpending=('TotalSpent', 'mean'),
    AverageQuantity=('TotalQuantity', 'mean'),
    AverageInvoices=('TotalInvoices', 'mean'),
    AverageOrderValue=('AverageOrderValue', 'mean')
).round(2)

cluster_summary
```

```python
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=customer_features,
    x='TotalSpent',
    y='TotalInvoices',
    hue='Cluster',
    palette='viridis',
    alpha=0.7
)

plt.title('Customer Segmentation Using K-Means')
plt.xlabel('Total Spending (£)')
plt.ylabel('Number of Invoices')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

df_ml = df_clean.copy()

df_ml['InvoiceDate'] = pd.to_datetime(df_ml['InvoiceDate'])
df_ml['InvoiceNo'] = df_ml['InvoiceNo'].astype(str)

# Keep rows with known customers
df_ml = df_ml.dropna(subset=['CustomerID']).copy()

df_ml['CustomerID'] = df_ml['CustomerID'].astype(int)

# Keep completed positive-quantity, positive-price sales
df_ml = df_ml[
    (~df_ml['InvoiceNo'].str.startswith('C')) &
    (df_ml['Quantity'] > 0) &
    (df_ml['UnitPrice'] > 0)
].copy()

df_ml['Revenue'] = df_ml['Quantity'] * df_ml['UnitPrice']

print(df_ml.shape)
df_ml.head()
```

```python
start_date = (
    df_ml['InvoiceDate'].min().normalize()
    + pd.Timedelta(days=90)
)

end_date = (
    df_ml['InvoiceDate'].max().normalize()
    - pd.Timedelta(days=30)
)

snapshot_dates = pd.date_range(
    start=start_date,
    end=end_date,
    freq='30D'
)

print("Number of snapshot dates:", len(snapshot_dates))
print("First snapshot:", start_date)
print("Last snapshot:", end_date)
```

```python
samples = []

for snapshot in snapshot_dates:

    # Transactions before the snapshot date
    history = df_ml[
        df_ml['InvoiceDate'] < snapshot
    ].copy()

    # Transactions during the following 30 days
    future = df_ml[
        (df_ml['InvoiceDate'] >= snapshot) &
        (df_ml['InvoiceDate'] < snapshot + pd.Timedelta(days=30))
    ]

    # Customer IDs with a purchase in the future window
    future_customers = set(future['CustomerID'].unique())

    # Historical customer-level features
    features = history.groupby('CustomerID').agg(
        TotalInvoices=('InvoiceNo', 'nunique'),
        TotalQuantity=('Quantity', 'sum'),
        TotalSpent=('Revenue', 'sum'),
        AverageUnitPrice=('UnitPrice', 'mean'),
        UniqueProducts=('StockCode', 'nunique'),
        LastPurchase=('InvoiceDate', 'max')
    )

    features['RecencyDays'] = (
        snapshot - features['LastPurchase']
    ).dt.days

    features = features.drop(columns='LastPurchase')

    # Create the target label
    features['PurchasedNext30Days'] = (
        features.index.isin(future_customers).astype(int)
    )

    features['SnapshotDate'] = snapshot

    samples.append(features.reset_index())

ml_data = pd.concat(samples, ignore_index=True)

print("Dataset shape:", ml_data.shape)
ml_data.head()
```

```python
dates = sorted(ml_data['SnapshotDate'].unique())

split_index = int(len(dates) * 0.8)

train_dates = dates[:split_index]
test_dates = dates[split_index:]

train_data = ml_data[
    ml_data['SnapshotDate'].isin(train_dates)
].copy()

test_data = ml_data[
    ml_data['SnapshotDate'].isin(test_dates)
].copy()

print("Training rows:", len(train_data))
print("Testing rows:", len(test_data))
print("Training period:", train_dates[0], "to", train_dates[-1])
print("Testing period:", test_dates[0], "to", test_dates[-1])
```

```python
feature_cols = [
    'RecencyDays',
    'TotalInvoices',
    'TotalQuantity',
    'TotalSpent',
    'AverageUnitPrice',
    'UniqueProducts'
]

X_train = train_data[feature_cols]
y_train = train_data['PurchasedNext30Days']

X_test = test_data[feature_cols]
y_test = test_data['PurchasedNext30Days']

print("Training target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

lr_model = make_pipeline(
    SimpleImputer(strategy='median'),
    StandardScaler(),
    LogisticRegression(
        class_weight='balanced',
        max_iter=1000,
        random_state=42
    )
)

lr_model.fit(X_train, y_train)

lr_predictions = lr_model.predict(X_test)
lr_probabilities = lr_model.predict_proba(X_test)[:, 1]

print("Logistic Regression training complete.")
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = make_pipeline(
    SimpleImputer(strategy='median'),
    RandomForestClassifier(
        n_estimators=200,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    )
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]

print("Random Forest training complete.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

def evaluate_model(name, y_true, predictions, probabilities):
    print(f"\n{name}")
    print("-" * 40)

    print("Accuracy:", round(
        accuracy_score(y_true, predictions), 3
    ))

    print("Precision:", round(
        precision_score(y_true, predictions, zero_division=0), 3
    ))

    print("Recall:", round(
        recall_score(y_true, predictions, zero_division=0), 3
    ))

    print("F1-score:", round(
        f1_score(y_true, predictions, zero_division=0), 3
    ))

    if y_true.nunique() == 2:
        print("ROC-AUC:", round(
            roc_auc_score(y_true, probabilities), 3
        ))
    else:
        print("ROC-AUC: unavailable; test data contains only one class")

    print("\nClassification report:")
    print(classification_report(
        y_true,
        predictions,
        zero_division=0
    ))

    print("Confusion matrix:")
    print(confusion_matrix(y_true, predictions))


evaluate_model(
    "Logistic Regression",
    y_test,
    lr_predictions,
    lr_probabilities
)

evaluate_model(
    "Random Forest",
    y_test,
    rf_predictions,
    rf_probabilities
)
```

```python
rf_classifier = rf_model.named_steps['randomforestclassifier']

importance = pd.Series(
    rf_classifier.feature_importances_,
    index=feature_cols
).sort_values(ascending=False)

print(importance)

importance.plot(kind='bar', figsize=(9, 5))
plt.title('Random Forest — Feature Importance')
plt.ylabel('Importance')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
results = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest'],
    'Accuracy': [
        accuracy_score(y_test, lr_predictions),
        accuracy_score(y_test, rf_predictions)
    ],
    'Precision': [
        precision_score(y_test, lr_predictions, zero_division=0),
        precision_score(y_test, rf_predictions, zero_division=0)
    ],
    'Recall': [
        recall_score(y_test, lr_predictions, zero_division=0),
        recall_score(y_test, rf_predictions, zero_division=0)
    ],
    'F1_Score': [
        f1_score(y_test, lr_predictions, zero_division=0),
        f1_score(y_test, rf_predictions, zero_division=0)
    ]
})

results.round(3)
```

```python
test_results = test_data[
    ['CustomerID', 'SnapshotDate', 'PurchasedNext30Days']
].copy()

test_results['LR_Prediction'] = lr_predictions
test_results['LR_Probability'] = lr_probabilities

test_results['RF_Prediction'] = rf_predictions
test_results['RF_Probability'] = rf_probabilities

test_results.to_csv(
    'purchase_prediction_results.csv',
    index=False
)

results.to_csv(
    'model_comparison.csv',
    index=False
)

print("Prediction results saved successfully.")
```
